```python
def count_ways(target, numbers):
    dp = [0] * (target + 1)
    dp[0] = 1
    for num in numbers:
        for i in range(num, target + 1):
            dp[i] += dp[i - num]
    return dp[target]


def count_ways_plusminus(target, numbers):
    dp = [0] * (target + 1)
    dp[0] = 1
    for num in numbers:
        for i in range(num, target + 1):
            dp[i] += dp[i - num]
    return dp[target]



numbers = [2, 4, 7, 9, 10, 12, 13, 15]
target = 148
print(count_ways(target, numbers))

numbers = list(range(1, 71))
print(count_ways(71, numbers))

numbers = list(range(1, 35))
print(count_ways(35, numbers))
```

```python
from collections import deque

def two_coloring(graph_data):
    num_vertices, edges = graph_data
    colors = [-1] * num_vertices  # -1 means uncolored
    adj_list = [[] for _ in range(num_vertices)]
    
    # Build adjacency list
    for u, v in edges:
        adj_list[u].append(v)
        adj_list[v].append(u)
    
    # Try to color the graph using BFS
    for start in range(num_vertices):
        if colors[start] == -1:  # If the vertex is not colored
            queue = deque([start])
            colors[start] = 0  # Start coloring with 0
            
            while queue:
                node = queue.popleft()
                current_color = colors[node]
                
                for neighbor in adj_list[node]:
                    if colors[neighbor] == -1:  # If the neighbor is not colored
                        colors[neighbor] = 1 - current_color  # Assign the opposite color
                        queue.append(neighbor)
                    elif colors[neighbor] == current_color:  # If the neighbor has the same color
                        return []  # Not possible to 2-color the graph
    
    return colors

graph_data = [11, [[8, 10], [0, 9], [5, 6], [7, 10], [6, 8], [0, 1], [5, 9], [4, 5], [1, 7], [1, 5], [3, 9], [7, 9], [3, 7], [0, 4], [0, 10], [1, 8], [5, 10]]]
print(two_coloring(graph_data))


```
